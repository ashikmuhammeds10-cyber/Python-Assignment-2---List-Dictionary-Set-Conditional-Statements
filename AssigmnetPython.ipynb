```python
age_list.insert(2, 30)
print(age_list)
```

```python
name_list.remove("Yazhini")
print(name_list)
```

```python
removed = age_list.pop()
print("Popped element:", removed)
print(age_list)
```

```python
age_list.extend([29, 30, 26])
print(age_list)
```

```python
age_list.sort(reverse=True)
print(age_list)
```

```python
my_set = set(['a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'])
print(my_set)
```

The list had 8 elements, but the set has only 5. A set stores only unique values, so the duplicate elements 'a' and 'i' were removed automatically. The output order also differs from the original list because sets are unordered.


```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
```

```python
print("Union:", set1.union(set2))
print("Intersection:", set1.intersection(set2))
```

```python
my_set[4] = 's'
```

Sets are unordered collections, so their elements have no index or position. Because of this, item assignment like my_set[4] = 's' is not allowed, and Python raises a TypeError. To change a set, use methods such as add(), remove() or discard().
